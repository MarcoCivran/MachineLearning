```python
import pandas as pd
import sklearn as sk

df = pd.read_csv("data\\DataCoSupplyChainDataset.csv", encoding="latin-1")
```

## Removing the Column

we want to keep only the rows which are not "Advance shipping" or "Shipping canceled" because they aren't meaningful for our objective <br>
Perfect, now we have a clean learning dataset!

```python
df = df[~df.iloc[:, 5].isin(["Advance shipping", "Shipping canceled"])].copy()

X = df.iloc[:, [9, 10, 25, 26, 28]]  # J, K, Z, AA, AC
Y = (df.iloc[:, 5] == "Late delivery").astype(int)  # F: late=1, on time=0
```

## Models

We start from separating learning set and test set, then we create a model for each thingy

```python
X_train, X_test, Y_train, Y_test = sk.model_selection.train_test_split(X, Y, test_size=0.2, random_state=42)

model_rf = sk.ensemble.RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42).fit(X_train, Y_train)
model_svm = sk.svm.SVC(random_state=42).fit(X_train, Y_train)
model_knn = sk.neighbors.KNeighborsClassifier(n_neighbors=5).fit(X_train, Y_train)
model_nb = sk.naive_bayes.GaussianNB().fit(X_train, Y_train)
```
