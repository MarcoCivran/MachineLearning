### Lab 1 - Hardest variable in Iris ###

1. Import the Iris dataset and split it into train and test datas

We choose to use the cross fold validation since the dataset is small and we want to have many "tries"

- for the "species" column (classification problem) you generate the accuracy compute the mean for each fold
- for the other columns (regression problem) you generate the R2 = 1 - (model MSE)/(baseline MSE) and compute the mean for each fold



```python
from sklearn.datasets import load_iris
import pandas as pd
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from sklearn.model_selection import RepeatedKFold, RepeatedStratifiedKFold, cross_val_score

iris = load_iris(as_frame=True)
df = iris.frame.drop(columns="target")
df["species"] = pd.Categorical.from_codes(iris.target, iris.target_names)
```

Compute the assessments of all columns:

```python
res = {}
for col in df.columns:
    X, y = pd.get_dummies(df.drop(columns=col)), df[col]
    cv = RepeatedKFold(n_splits=5, n_repeats=10, random_state=0)
    if col == "species":
        res[col] = cross_val_score(
            DecisionTreeClassifier(min_samples_leaf=5, random_state=0),
            X, y, cv=cv, scoring="accuracy").mean()
    else:
        res[col] = cross_val_score(
            DecisionTreeRegressor(min_samples_leaf=5, random_state=0),
            X, y, cv=cv, scoring="r2").mean()

print(pd.Series(res).sort_values())
```
