### Initiate population

```python
import random
import string

n_pop = 50
target = ''.join(random.choices(string.ascii_letters, k=random.randint(3, 20) ))
print("Target:", target)
```

## Implementation
Mutation, fitness, selection

```python
import Levenshtein

def mutation(individual, mutation_rate):
    result = ''

    for character in individual:
        if (random.random() < mutation_rate):
            new_character = random.choice(string.ascii_letters)
            result += new_character
        else:
            result += character
        if (random.random() < mutation_rate):
            new_character = random.choice(string.ascii_letters)
            result += new_character
        if (random.random() < mutation_rate):
            result = result[:-1]
    return result

def fitness(individual):
    return Levenshtein.distance(individual, target)



def tournament_selection(population, tournament_size, fit):
    selected = random.sample(population, tournament_size)
    return min(selected, key=fit)

def generation(pop, fit, mutation, mut_rate, k_tournament):
    new_pop = []
    for i in range(len(pop)):
        selected = tournament_selection(pop, k_tournament, fit)
        mutated = mutation(selected, mut_rate)
        new_pop.append(min([selected, mutated], key=fit))
    return new_pop
```

Now we must implement the EA

```python
def e_a_sport(pop_size, target, k_tournament, mut_rate, mutation, fit, n_gen):
    # pop initialization    
    pop = []
    hist = []

    for i in range(n_pop):
        length = random.randint(3, 20) 
        random_string = ''.join(random.choices(string.ascii_letters, k=length))
        pop.append(random_string)

    for i in range(n_gen):
        curr_best = min(pop, key=fit)
        hist.append(fit(curr_best))
        pop = generation(pop, fit, mutation, mut_rate, k_tournament)

    curr_best = min(pop, key=fit)
    hist.append(fit(curr_best))

    return curr_best, hist
```

```python
print(target)
print(e_a_sport(pop_size=100, target=target, k_tournament=10, mut_rate=0.1, mutation=mutation, fit=fitness, n_gen=1000))
```
